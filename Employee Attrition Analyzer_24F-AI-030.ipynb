# Preprocessing

```python
#I drop it as it doesnot affect my output because its only a single value '1'
df.drop(columns='EmployeeCount',axis=1,inplace=True)
```

```python
#I drop it as it also dont affect my output, like how can someone says that your age is greater so you dont leave out company
df.drop(columns='Over18',axis=1,inplace=True)
```

```python
#I drop it as it doesnot affect my output because its only a single value '80'.
df.drop(columns='StandardHours',axis=1,inplace=True)
```

```python
df.head()
```

```python
#Doing Encoding like converting the categorical data into numeric as our model only understands numerical data
```

```python
df.select_dtypes(include='object').head()
```

```python
X = df.drop(columns='Attrition',axis=1)
y = df['Attrition']
```

```python
categorical_cols = X.select_dtypes(include='object').columns
```

```python
#using onehotencoding over categorical data of features
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
```

```python
X.head()
```

```python
#using label encoder over our target varaible to encode it
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
y = le.fit_transform(y)
```

```python
y
```

```python
#ok now splitting the data for testing as training
from sklearn.model_selection import train_test_split
```

```python
 X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
X_train.shape, X_test.shape
```

# Training The Models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
lr = LogisticRegression()
dt = DecisionTreeClassifier()
rf = DecisionTreeClassifier()
```

```python
lr.fit(X_train,y_train)
```

```python
dt.fit(X_train,y_train)
```

```python
rf.fit(X_train,y_train)
```

```python
lr_pred = lr.predict(X_test)
dt_pred = dt.predict(X_test)
rf_pred = rf.predict(X_test)
```

# Evaluating the Model & Selecting the Best

```python
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
```

```python
accuracy_score(y_test,lr_pred)
```

```python
lr_train_pred = lr.predict(X_train)
accuracy_score(y_train,lr_train_pred)
```

```python
accuracy_score(y_test,dt_pred)
```

```python
accuracy_score(y_test,rf_pred)
```

```python
print(classification_report(y_test,lr_pred))
```

```python
print(classification_report(y_test,dt_pred))
```

```python
print(classification_report(y_test,rf_pred))
```

```python
sns.heatmap(confusion_matrix(y_test,lr_pred),annot=True,fmt='g')
plt.show()
```

```python
sns.heatmap(confusion_matrix(y_test,dt_pred),annot=True,fmt='g')
plt.show()
```

```python
sns.heatmap(confusion_matrix(y_test,rf_pred),annot=True,fmt='g')
plt.show()
```
